## Scope and assumptions

Every request in this project is synthetic and generated with a fixed seed. The model tiers and token prices are illustrative scenario inputs, not current provider prices. Failed requests are assigned zero billable tokens for this exercise. The analysis demonstrates a workflow; it does not describe production usage or prove causal savings.

```python
from pathlib import Path
import pandas as pd
from IPython.display import SVG, Markdown, display

ROOT = Path.cwd()
if not (ROOT / 'reports' / 'findings.md').exists():
    ROOT = ROOT.parent

display(Markdown((ROOT / 'reports' / 'findings.md').read_text(encoding='utf-8')))

display(Markdown('## Spend by team and model tier'))
display(pd.read_csv(ROOT / 'reports' / 'cost_by_team_model.csv').head(12))
display(SVG(filename=str(ROOT / 'reports' / 'spend_by_model.svg')))

display(Markdown('## Daily reliability'))
display(pd.read_csv(ROOT / 'reports' / 'daily_operations.csv').head(10))
display(SVG(filename=str(ROOT / 'reports' / 'daily_error_rate.svg')))

display(Markdown('## Highest failure groups'))
display(pd.read_csv(ROOT / 'reports' / 'failure_ranking.csv').head(10))
```

## How to interpret this

Use the charts to locate groups that may deserve investigation, then verify the result against real billing and reliability definitions before acting. The sample is intentionally simple: it does not model retries, partial billing, provider-specific pricing rules, or real user behavior.
